## 1. Create 1D, 2D and 3D arrays

```python
import numpy as np
one_d = np.array([10, 20, 30, 40])
two_d = np.array([[1, 2, 3], [4, 5, 6]])
three_d = np.arange(24).reshape(2, 3, 4)
print(one_d, two_d, three_d, sep="\n\n")
```

## 2. Vectorised arithmetic and aggregation across axes

```python
values = np.array([[10, 20, 30], [40, 50, 60]])
print("Add 5:", values + 5)
print("Multiply by 2:", values * 2)
print("Column sums (axis=0):", values.sum(axis=0))
print("Row sums (axis=1):", values.sum(axis=1))
```

## 3. Boolean masking and fancy indexing

```python
arr = np.array([12, 25, 8, 40, 17, 55])
print("Values above 20:", arr[arr > 20])
print("Fancy indexing positions 0, 2, 5:", arr[[0, 2, 5]])
```

## 4. Statistics and correlation

```python
data = np.array([12, 15, 18, 20, 22, 25, 30], dtype=float)
print("Mean:", np.mean(data))
print("Standard deviation:", np.std(data))
print("90th percentile:", np.percentile(data, 90))
print("Correlation matrix:\n", np.corrcoef(data, data * 2 + 3))
```

## 5. Python loop versus NumPy vectorisation — one million values

```python
import time
n = 1_000_000
py_values = list(range(n))
start = time.perf_counter()
py_result = [x * 2 + 1 for x in py_values]
python_seconds = time.perf_counter() - start
np_values = np.arange(n)
start = time.perf_counter()
np_result = np_values * 2 + 1
numpy_seconds = time.perf_counter() - start
print(f"Python list comprehension: {python_seconds:.6f} seconds")
print(f"NumPy vectorisation:       {numpy_seconds:.6f} seconds")
print("Speed ratio (Python/NumPy):", python_seconds / numpy_seconds if numpy_seconds else "N/A")
```

Benchmark results depend on hardware, Python version, and system load. Run more than once for a more stable comparison.
