## 1. Load and inspect the operational dataset
The primary dataset contains hospital-level resource utilisation records.

```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

ROOT = Path(__file__).resolve().parent
# If running from another directory, adjust ROOT to your repository root.
DATA = ROOT / "data" / "week-2" / "raw"
OUT = ROOT / "reports" / "week-2"
OUT.mkdir(parents=True, exist_ok=True)
df = pd.read_csv(DATA / "sunrise_resource_utilisation.csv")
df["date"] = pd.to_datetime(df["date"], errors="coerce")
sns.set_theme(style="whitegrid", palette="deep")
display(df.head())
print("Shape:", df.shape)
```

## 2. Line chart — bed occupancy over time
Daily average occupancy percentage is used where dates are available.

```python
daily = df.assign(occupancy_pct=100 * df["occupied_beds"] / df["total_beds"].replace(0, np.nan))
daily = daily.groupby("date", as_index=False)["occupancy_pct"].mean().sort_values("date")
plt.figure(figsize=(11, 5))
sns.lineplot(data=daily, x="date", y="occupancy_pct", marker="o")
plt.title("Average Hospital Bed Occupancy Over Time")
plt.xlabel("Date"); plt.ylabel("Occupancy (%)"); plt.xticks(rotation=45); plt.tight_layout()
plt.savefig(OUT / "bed_occupancy_line.png", dpi=150); plt.show()
```

## 3. Bar chart — average waiting time by department

```python
dept_wait = df.groupby("department", as_index=False)["avg_patient_wait_mins"].mean().sort_values("avg_patient_wait_mins", ascending=False)
plt.figure(figsize=(10, 5))
sns.barplot(data=dept_wait, x="department", y="avg_patient_wait_mins")
plt.title("Average Patient Waiting Time by Department")
plt.xlabel("Department"); plt.ylabel("Average wait (minutes)"); plt.xticks(rotation=35); plt.tight_layout()
plt.savefig(OUT / "department_wait_bar.png", dpi=150); plt.show()
```

## 4. Scatter chart — bed capacity and occupied beds

```python
plt.figure(figsize=(7, 5))
sns.scatterplot(data=df, x="total_beds", y="occupied_beds", hue="hospital")
plt.title("Total Beds vs Occupied Beds"); plt.xlabel("Total beds"); plt.ylabel("Occupied beds")
plt.tight_layout(); plt.savefig(OUT / "beds_scatter.png", dpi=150); plt.show()
```

## 5. Histogram — waiting time distribution

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="avg_patient_wait_mins", bins=25, kde=True)
plt.title("Distribution of Average Patient Waiting Time"); plt.xlabel("Waiting time (minutes)")
plt.tight_layout(); plt.savefig(OUT / "waiting_histogram.png", dpi=150); plt.show()
```

## 6. Box plot — waiting time by department

```python
plt.figure(figsize=(10, 5))
sns.boxplot(data=df, x="department", y="avg_patient_wait_mins")
plt.title("Waiting Time Spread and Potential Outliers by Department")
plt.xlabel("Department"); plt.ylabel("Waiting time (minutes)"); plt.xticks(rotation=35)
plt.tight_layout(); plt.savefig(OUT / "waiting_boxplot.png", dpi=150); plt.show()
```

## Interpretation
Use the charts to compare departments, identify variation, and investigate unusual values. Do not treat every statistical outlier as an error.
